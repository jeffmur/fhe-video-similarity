This notebook serves to generate F1 Score, Precision, Recall, and Accuracy to compare to the results of the supervised learning models from SSO.

## Field of View (FOV) Dataset

... methodology


The field of view degrees are calculated counter-clockwise. Where 0 degrees is in front of the subject, 90 degrees is left of the subject, 180 degrees is behind the subject, and 270 degrees is right of the subject.

As an additional note, some phones record in portrait or landscape mode. The dataset is comprised of 3 simultaneous, stationary mobile devices recording from various angles.


```python
from utils.ann import train_test_ann, dir_to_df, grid_search_ann, drawROC, confusionMatrix

LabelRecord = []
PredictRecord = []
ModelName = []
```

### Experiment 1: Low Movement

Recording from various angles of myself for 20 minutes at my desk listening to music.

Google Pixel 3XL - Landscape - 90 degrees

iPhone 13 - Landscape - 180 degrees

Samsung Galaxy S9 - Portrait - 270 degrees

```python
y_test, y_pred, y_pred_proba =  train_test_ann(dir_to_df('2_fov/1_office/train_similarity_scores')).values()
confusionMatrix(y_test, y_pred)
LabelRecord.append(y_test)
PredictRecord.append(y_pred)
ModelName.append('ANN')
```

```python
y_test, y_pred, y_pred_proba = grid_search_ann(dir_to_df('2_fov/1_office/train_similarity_scores')).values()
confusionMatrix(y_test, y_pred)
LabelRecord.append(y_pred)
PredictRecord.append(y_pred_proba)
ModelName.append('ANN_GridSearch')
```

```python
drawROC(LabelRecord, PredictRecord, ModelName)
```
